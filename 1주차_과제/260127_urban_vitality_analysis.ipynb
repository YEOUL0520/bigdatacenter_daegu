1. 데이터 구성 및 기본 정보 확인
- daegu_service_pcell_time csv를 활용
- 컬럼명, 상위 5개, 하위 5개, info 정보를 확인

```python
import os
import pandas as pd
from IPython.display import display

DATA_DIR = "."
files = ["daegu_service_pcell_time_pop_202510.csv"]

for f in files:
    path = os.path.join(DATA_DIR, f)
    print(f, "=>", "Ok" if os.path.exists(path) else "Not FOUND")

def inspect_csv(path, nrows_preview=2000):
    print("\n"+"="*90)
    print(f"[FILE] {os.path.basename(path)}")

    encodings_to_try = [None, "utf-8", "cp949"]
    df_preview = None
    used_encoding = None
    last_err = None

    for enc in encodings_to_try:
        try:
            df_preview = pd.read_csv(path, nrows=nrows_preview, encoding=enc)
            used_encoding = enc
            break
        except UnicodeDecodeError as e:
            raise last_err
    
    if df_preview.shape[1] == 1 and "|" in str(df_preview.columns[0]):
        df_preview = pd.read_csv(path, nrows=nrows_preview, encoding=None if used_encoding=="default" else used_encoding, sep = "|")
        print("Detected '|' delimeter -> reloaded with sep'|'")

    enc_label = "default" if used_encoding is None else used_encoding

    print(f"Encoding used: {used_encoding}")
    print(f"Preview shape: {df_preview.shape}")

    print("\n[Columns]")
    print(list(df_preview.columns))

    print("\n[Head_5]")
    display(df_preview.head(5))

    print("\n[Tail_5]")
    display(df_preview.tail(5))

    print("\n[Info]")
    df_preview.info()

    return df_preview

preview_dfs = {}
for f in files:
    path = os.path.join(DATA_DIR, f)
    if os.path.exists(path):
        preview_dfs[f] = inspect_csv(path)
```

2. 시간대별 평균 컬럼 생성
- 대구 전체에 대한 통계
- 07-11 / 12-18 / 19-22 / 23-06 시간대별 통계 산출

```python
import pandas as pd
import numpy as np

FILE = "daegu_service_pcell_time_pop_202510.csv"

TIME_BANDS = {
    "07_11" : list(range(7,12)),
    "12_18" : list(range(12,19)),
    "19_22" : list(range(19,23)),
    "23_06" : [23, 0, 1, 2, 3, 4, 5, 6],
}
bands = list(TIME_BANDS.keys())

base_cols = ["STD_YMD", "HCODE", "X_COORD", "Y_COORD"]
H_cols = [f"H_T_{h:02d}" for h in range(24)]
W_cols = [f"W_T_{h:02d}" for h in range(24)]
V_cols = [f"V_T_{h:02d}" for h in range(24)]
usecols = base_cols + H_cols + W_cols + V_cols

dtypes = {
    "STD_YMD" : "int64",
    "HCODE" : "int64",
    "X_COORD" : "float64",
    "Y_COORD" : "float64",
}
for c in H_cols + W_cols + V_cols:
    dtypes[c] = "float32"

df = pd.read_csv(FILE, sep="|", usecols=usecols, dtype=dtypes)
print(df.shape)

def add_band_means(df: pd.DataFrame, prefix: str):
    for band, hours in TIME_BANDS.items():
        cols = [f"{prefix}_T_{h:02d}" for h in hours]
        cols = [c for c in cols if c in df.columns]
        df[f"{prefix}_B_{band}"] = df[cols].mean(axis=1)
    return df

df = add_band_means(df, "H") #거주인구
df = add_band_means(df, "W") #직장인구
df = add_band_means(df, "V") #방문인구

for b in bands:
    df[f"ALL_B_{b}"] = df[f"H_B_{b}"] +df[f"W_B_{b}"] +df[f"V_B_{b}"]

ordered = []
for b in bands:
    ordered += [f"H_B_{b}", f"W_B_{b}", f"V_B_{b}", f"ALL_B_{b}"]

daegu_overall_mean = df[ordered].mean().to_frame("DAEGU_OVERALL_MEAN")
display(daegu_overall_mean.round(3))
```

3. 지역별 통계 컬럼 생성
- 원본 csv와의 HCODE 매핑을 통해 구/군 매칭
- 오류 확인을 위한 결측 비율 확인

```python
def map_sigungu_from_hcode(hcode):
    hcode = str(hcode)
    if hcode.startswith("27110"):
        return "중구"
    elif hcode.startswith("27140"):
        return "동구"
    elif hcode.startswith("27170"):
        return "서구"
    elif hcode.startswith("27200"):
        return "남구"
    elif hcode.startswith("27230"):
        return "북구"
    elif hcode.startswith("27260"):
        return "수성구"
    elif hcode.startswith("27290"):
        return "달서구"
    elif hcode.startswith("27710"):
        return "달성군"
    else:
        return None
    
df["SIGUNGU_NM"] = df["HCODE"].apply(map_sigungu_from_hcode)
df["SIGUNGU_NM"].value_counts()

bands = list(TIME_BANDS.keys())

ordered = []
for b in bands:
    ordered += [
        f"H_B_{b}",
        f"W_B_{b}",
        f"V_B_{b}",
        f"ALL_B_{b}"
    ]

gu_mean = (
    df.groupby("SIGUNGU_NM")[ordered]
    .mean()
    .round(3)
    .reset_index()
)

display(gu_mean)
print("gu_mean shape: ", gu_mean.shape)

print("결측비율확인:", df["SIGUNGU_NM"].isna().mean())
print(df["SIGUNGU_NM"].value_counts(dropna=False).head(20))

print(df["HCODE"].head(10).tolist())
print(df["HCODE"].dtype)
```

4. 통계 결과 저장

```python
pcell_cols = [
    "X_COORD", "Y_COORD", "HCODE", "SIGUNGU_NM",
    "H_B_07_11", "H_B_12_18", "H_B_19_22", "H_B_23_06",
    "W_B_07_11", "W_B_12_18", "W_B_19_22", "W_B_23_06",
    "V_B_07_11", "V_B_12_18", "V_B_19_22", "V_B_23_06",
    "ALL_B_07_11", "ALL_B_12_18", "ALL_B_19_22", "ALL_B_23_06"
]

df_pcell = df[pcell_cols]

df_pcell.to_csv(
    "daegu_pcell_timeband_202510.csv",
    index=False,
    encoding="utf-8-sig"
)
```

5. 그리드 생성
- 좌표를 500m 단위 버킷팅
- 같은 격자에 속한 값 평균
- 결과 -> 격자 1칸 = 1행

```python
GRID_SIZE = 500

df["GX"] = (df["X_COORD"] // GRID_SIZE) * GRID_SIZE
df["GY"] = (df["Y_COORD"] // GRID_SIZE) * GRID_SIZE

grid_mean = (
    df.groupby(["GX", "GY"])
    .agg({
        "ALL_B_07_11": "mean",
        "ALL_B_12_18": "mean",
        "ALL_B_19_22": "mean",
        "ALL_B_23_06": "mean",
    })
    .reset_index()
)

grid_mean.head()
```

```python
grid_mean.to_csv(
    "daegu_grid_500m_mean.csv",
    index = False,
    encoding = "utf-8"
)
```
